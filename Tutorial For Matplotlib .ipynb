```python
import matplotlib.pyplot as plt
import numpy as np
```

## Line plots

```python
x = [1, 2, 4, 5, 6]
y = [2, 4, 8, 10, 12]
x2 = [1,4,16,25,16]
plt.figure(figsize=(8,6))

plt.plot(x,y,label="y=2x",color="blue",linewidth=2,marker="*",linestyle="--")
plt.plot(x,x2,label="x2=x**2",color="green",linewidth=3,marker="o",linestyle="--")

plt.title("A line Graph", fontdict={"fontname" : "Comic Sans MS" , "fontsize" : 24})

plt.xlabel("time",fontdict={"fontname" : "Comic Sans MS" , "fontsize" : 24})
plt.ylabel("price",fontdict={"fontname" : "Comic Sans MS" , "fontsize" : 24})

plt.xticks(np.arange(0,7,0.5))# the range of x axis 
plt.yticks(np.arange(0,30,5))# the range of y axis 

plt.savefig("graph.png")#"graph.pdf",("graph.svg", dpi= 500 )#dpi = (dots per inch ) ,

plt.legend(loc=2)
plt.show()
```

```python
#make four graphs(2, 2)
x = np.arange(100)

plt.figure(figsize=(20, 12))
fig , axs = plt.subplots(2, 2)

axs[0, 0].plot(x, np.sin(x))
axs[0, 0].set_title("Sin wave ")

axs[0, 1].plot(x, np.cos(x))
axs[0, 1].set_title("Cos wave ")

axs[1, 0].plot(x, np.log(x))
axs[1, 0].set_title("Log wave ")

axs[1, 1].plot(x, np.random.random(100))
axs[1, 1].set_title("Random wave ")

plt.show()

```

## Bar chart plots

```python
lables = ["A", "B", "C"] 
values = [1,2,4]

plt.figure(figsize=(8,6))
plt.bar(lables,values,color="green",width=.2)

plt.title("Bar Plot")
plt.xlabel("Labels")
plt.ylabel("Values")

plt.xticks(rotation=45)
plt.yticks(np.arange(0,6,2))

plt.show()
```

```python
lables = ["A", "B", "C"] 
values = [1,2,4]

plt.figure(figsize=(8,6))
bar_colors=["red","blue","green"]
bars=plt.bar(lables,values,color=bar_colors,width=.2)
bars[2].set_hatch("/")

plt.title("Bar Plot")
plt.xlabel("Labels")
plt.ylabel("Values")

plt.xticks(rotation=45)
plt.yticks(np.arange(0,6,2))

plt.show()
```

```python
company =["Google", "AMZN", "MSFT", "FB"] #X_axis
revenue =[90,136,85,27] # Y_axis
profit =[40,7,34,12] # secondry y_axis 

plt.figure(figsize=(8,6))
plt.bar(company, revenue, width=0.4 , label= "Revenue")
plt.bar(company, profit, width=0.4 , label= "Profit")

plt.ylabel("Revenue(Bln)")
plt.title("US Technology Stocks")
plt.legend()
plt.show()
```

## Scatter chart plots¶

## Pie chart plots

```python
exp_vals = [1400,600,300,300,250]
exp_labels= ["home rent", "food", "internet/phone","car", "other"]

plt.figure(figsize=(8,6))
plt.pie(exp_vals,labels=exp_labels,autopct="%1.2f%%",radius=.8,explode=[.2,0,0,0,0])#explode = you should put in it the value of distance which you want the pie go out from figure
plt.title("Pie chart")
plt.legend(bbox_to_anchor=(1, 1))

plt.show()
```

## 3D plots

```python
axis = plt.axes(projection = "3d")
x = np.arange(0, 50, 0.1)
y = np.sin(x)
z = np.cos(x)

plt.figure(figsize=(6,6))
axis.plot(x, y, z)
axis.set_title("3D plot ")
axis.set_xlabel("test")

plt.show()
```

```python
axis = plt.axes(projection = "3d")
x = np.arange(-5, 5, 0.1)
y = np.arange(-5, 5, 0.1)
X,Y = np.meshgrid(x,y)
Z = np.sin(X) * np.cos(Y)


axis.plot_surface(X, Y, Z)
axis.set_title("3D plot ")
axis.set_xlabel("test")
plt.show()
```
