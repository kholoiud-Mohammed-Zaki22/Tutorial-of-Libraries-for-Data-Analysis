```python
tips=sns.load_dataset("tips")
tips
```

## Set Style
### ('darkgrid', 'whitegrid', 'dark', 'white', 'ticks')
### sns.set_style('whitegrid')

## Scatter Plot

```python

sns.set_style('darkgrid')#'whitegrid', 'dark', 'white', 'ticks'
sns.scatterplot(x="total_bill",y="tip",data=tips,hue="sex",size="size",palette="magma")#palette="coolwarm",,palette="viridis",,palette="magma",,palette="Set1"

plt.title("Scatter Plot")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.xticks(np.arange(0,55,5))
plt.yticks(np.arange(0,12,2))
plt.show()
```

## Histogram Plot

```python
#histographs (kde is the line if you want to put it or no)(bins = number of bins you want )
sns.set_style('darkgrid')#'whitegrid', 'dark', 'white', 'ticks'
sns.histplot(tips["tip"],bins=20,kde=True,color="blue")

plt.title("Histogram")
plt.xlabel("Freduency")
plt.ylabel("Tip")

plt.show()
```

## Distripution plot

```python
sns.set_style('darkgrid')#'whitegrid', 'dark', 'white', 'ticks'
sns.displot(tips["tip"],bins=20,kde=True,color="blue")

plt.title("Distripution")
plt.xlabel("Freduency")
plt.ylabel("Tip")

plt.show()
```

## Bar Plot

```python

sns.set_style('darkgrid')#'whitegrid', 'dark', 'white', 'ticks'
sns.barplot(x="sex",y="total_bill",data=tips,hue="day",palette="magma",estimator=sum)#palette = "YlGnBu"

plt.title("Bar Plot")
plt.xlabel("Sex")
plt.ylabel("Total_Bill")

plt.show()
```

```python
sns.set_style('darkgrid')#'whitegrid', 'dark', 'white', 'ticks'
sns.barplot(x="total_bill",y="sex",data=tips,hue="day",palette="magma",estimator=sum)#palette = "YlGnBu"

plt.title("Bar Plot")
plt.xlabel("Sex")
plt.ylabel("Total_Bill")

plt.show()
```

## Box Plot

```python
sns.set_style('darkgrid')#'whitegrid', 'dark', 'white', 'ticks'
sns.boxplot(data=tips,x="tip",color="green")

plt.title("Box Plot")
plt.xlabel("Tip")


plt.show()
```

```python
sns.set_style('darkgrid')#'whitegrid', 'dark', 'white', 'ticks'
sns.boxplot(data=tips,x="sex",y="tip",color="green")

plt.title("Box Plot")
plt.xlabel("Tip")


plt.show()
```

```python
sns.set_style('darkgrid')#'whitegrid', 'dark', 'white', 'ticks'
sns.boxplot(data=tips,x="sex",y="tip",hue="sex",palette="magma")# palette="YlGnBu"

plt.title("Box Plot")
plt.xlabel("Tip")


plt.show()
```

## Regression Plot

```python
sns.set_style('darkgrid')#'whitegrid', 'dark', 'white', 'ticks'
sns.lmplot(x="total_bill",y="tip",data=tips, hue ='sex',scatter_kws={"s" : 25}, palette="magma", markers=["o", "v"])

plt.title("Regression Plot")
plt.xlabel("Total_Bil")
plt.ylabel("tip")

plt.show()
```

## Joint plots

```python
#reg prefer for a regretion line 
sns.set_style('darkgrid')#'whitegrid', 'dark', 'white', 'ticks'
sns.jointplot(x="total_bill", y="tip", data=tips, kind="reg",color="green")

plt.title("Joint plots")
plt.xlabel("Total_Bil")
plt.ylabel("tip")

plt.show()
```

## Pair plots
## Pair plots will plot scatterplots for numerical columns and histograms for the diagonal where a variable is plotted against itself. It's a quick way to visualize relationships between multiple variables in your dataset.¶

```python
sns.set_style('dark')#'whitegrid', 'dark', 'white', 'ticks'
sns.set_style('whitegrid')

sns.pairplot(tips,hue="sex")


plt.show()
```

## Heat map plot

```python
titanic=sns.load_dataset("titanic")
titanic
```

```python
matrix=titanic.corr(numeric_only=True)
sns.set_style('darkgrid')#'whitegrid', 'dark', 'white', 'ticks'
sns.heatmap(matrix,annot=True,cmap="coolwarm")

plt.title("Heatmap")

plt.show()
```
